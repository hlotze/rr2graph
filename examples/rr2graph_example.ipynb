# rr2graph - example session

best do this directly at the console, for documentation purposes its done here in a Jupyter notebook

- create new folder
- change to that new folder

```python
mkdir rr2graph_test
cd rr2graph_test
```

- install a venv

```python
%%bash
python -m venv venv
```

- activate that venv

```python
%%bash
source venv/bin/activate
```

- install rr2graph package from PyPI

```python
%pip install rr2graph
```

- call the help from rr2graph

```python
%%bash 
rr2graph -h
```

- call the version info

```python
%%bash
rr2graph -v
```

- get the --info 

```python
%%bash
rr2graph -i
```

- generate the test data with option -g to get the test_rr_data.xlsx in your directory

```python
%%bash
rr2graph -g
```

```python
%%bash
ls test_rr_data.xlsx
```

- take that test data and let rr2graph generate the plots at subdir plots/

```python
%%bash
rr2graph -e test_rr_data.xlsx -n 3 -o plots/
```

```python
%%bash
tree plots/
```

you can use config.yaml to shorten your command prompt

```python
%%bash
more config.yaml
```

```python
%%bash
rr2graph -c config.yaml
```

```python
%%bash
tree plots/
```
